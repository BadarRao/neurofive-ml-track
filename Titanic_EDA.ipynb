## Titanic Dataset - Initial Findings

The Titanic training dataset contains 891 passenger records and 12 features.
Several columns contain missing values, with Cabin having the highest number of missing entries, followed by Age and Embarked.
The dataset includes both numerical and categorical features such as Age, Fare, Sex, and Passenger Class.
The survival variable is binary and appears to be the target variable for future machine learning tasks.
Passenger demographics, ticket information, and travel class may influence survival outcomes.
Further data cleaning and visualization will be required before building predictive models.
